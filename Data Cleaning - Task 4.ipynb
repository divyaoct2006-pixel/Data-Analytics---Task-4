```python
def full_pipeline(file):
    import pandas as pd
    
    # Load data
    df = pd.read_excel(file)
    df = df.copy()
    
    # ---------------------------
    # 1. Handle Missing Values
    # ---------------------------
    df['discount_percentage'] = df['discount_percentage'].fillna(0)
    
    df['customer_satisfaction_score'] = df['customer_satisfaction_score'].fillna(
        df['customer_satisfaction_score'].median()
    )
    
    df['email_open_rate'] = df['email_open_rate'].fillna(
        df['email_open_rate'].median()
    )
    
    df['days_since_last_purchase'] = df['days_since_last_purchase'].fillna(999)
    
    # ---------------------------
    # 2. Fix Date
    # ---------------------------
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    
    # ---------------------------
    # 3. Clean Text Columns
    # ---------------------------
    cat_cols = df.select_dtypes(include='object').columns
    for col in cat_cols:
        df[col] = df[col].str.strip().str.lower()
    
    # ---------------------------
    # 4. Handle Outliers (Capping)
    # ---------------------------
    cols = ['sales_revenue_usd', 'website_traffic', 'marketing_budget_usd']
    
    for col in cols:
        lower = df[col].quantile(0.01)
        upper = df[col].quantile(0.99)
        df[col] = df[col].clip(lower, upper)
    
    # ---------------------------
    # 5. Feature Engineering
    # ---------------------------
    df['total_ad_spend'] = df['ad_spend_online_usd'] + df['ad_spend_offline_usd']
    
    df['roi'] = df['sales_revenue_usd'] / df['total_ad_spend']
    
    df['engagement'] = df['website_traffic'] * df['conversion_rate']
    
    df['year'] = df['date'].dt.year
    df['month'] = df['date'].dt.month
    
    # ---------------------------
    # 6. Fix ROI Issues
    # ---------------------------
    df['roi'] = df['roi'].replace([float('inf'), -float('inf')], 0)
    df['roi'] = df['roi'].clip(0, df['roi'].quantile(0.99))
    
    # ---------------------------
    # 7. Final Cleanup
    # ---------------------------
    df = df.drop_duplicates()
    
    return df
```

```python
df = full_pipeline("marketing_sales_dataset.xlsx")
```
